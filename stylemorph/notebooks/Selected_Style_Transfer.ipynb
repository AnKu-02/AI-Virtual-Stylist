🧠 Goal:
Apply the style only to the hoodie, while preserving the original face, pants, and shoes.



### Hoodie mask

```python
import cv2
import numpy as np
from PIL import Image

# Load image and convert
image_path = '../outputs/01masked_full.jpg'
image = cv2.imread(image_path)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Crop only torso region: (y1:y2, x1:x2) — adjust these manually!
torso_crop = image_rgb[300:700, 250:550]  # <-- These numbers may need tuning

# Gray range for hoodie
lower = np.array([40, 40, 40])
upper = np.array([180, 180, 180])

# Mask only on torso crop
torso_mask = cv2.inRange(torso_crop, lower, upper)

# Place torso mask back into full-size mask
mask = np.zeros(image_rgb.shape[:2], dtype=np.uint8)
mask[300:700, 250:550] = torso_mask

# Morph cleanup (optional)
kernel = np.ones((5, 5), np.uint8)
mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
mask = cv2.morphologyEx(mask, cv2.MORPH_DILATE, kernel)

# Save and preview
cv2.imwrite("../outputs/hoodie_mask_refined.png", mask)
Image.fromarray(mask).show()
```
